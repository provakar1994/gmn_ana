```python
dfs_all = load_datasets(file_groups)
```

```python
value_columns = ["R3", "R5", "R4", "R1", "R6", "R2", "R7"]
error_columns = [f"{column}err" for column in value_columns]
kin_order = [430, 450, 970, 870, 1470, 785, 11100]
mode_order = ["Prod", "OTm3", "NOeffic", "NOnblkHCALcut"]
```

```python
df_pivot = build_table(dfs_all, value_columns, kin_order, mode_order)
df_pivot_err = build_table(dfs_all, error_columns, kin_order, mode_order)
df_pivot_err["R"] = df_pivot_err["R"].str.replace("err", "", regex=False)
```
