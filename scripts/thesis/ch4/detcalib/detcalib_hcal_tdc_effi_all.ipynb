```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
q2 = [3,4.5,7.4,9.9,13.6]
effi = [0.96,0.94,0.80,0.77,0.40]
```

```python
df = pd.DataFrame(
    {'q2': q2,
     'effi': effi
    })
df.head()
```

```python
# plotting various parameters vs cut width
plt.style.use('default')
sns.set_context('talk')
fig, ax = plt.subplots()
#fig.set_size_inches(10, 8)

#ax.plot(xlab, df_wcut_scan['R1'], marker='o', linestyle='--', color='b')
ax.scatter(df['q2'], df['effi'], marker='o', s=40, color='r')

ax.set_ylim(0,1)
#ax.set_xlim(0,15)
ax.set_ylabel(r"HCAL TDC Efficiency")
ax.set_xlabel(r'Q$^2$ (GeV$^2$)')
ax.grid(axis = "y")
#ax.set_title('W cut scan (w/ fidu. cut) | SBS4-SBS50% data')

plt.xticks(q2)

plt.tight_layout()

plt.savefig("detcalib_hcal_tdc_effi_all.pdf")

plt.show()
```
