## This script is used to concatenate get_charge_from_Bobs_script_gmn_all_tg.csv with beamCh_n_daqLvTM_prun_Target.csv files

```python
import pandas as pd
```

### Step 1: Reading charge generated using Bob's script

```python
# Read sheet with Bob's charge numbers
bobch = pd.read_csv('get_charge_from_Bobs_script_gmn_all_tg.csv')
bobch.head()
```

```python
# check for nulls
anyNaN_bob = bobch.isnull().values.any()
#final_table_nz_nd.loc[pd.isna(final_table_nz_nd).any(1), :].index
#if (anyNaN == True): 
#    print('NaN exists! Replacing with 0')
#    concatdfch = concatdfch.fillna(0)
print(anyNaN_bob)
```

### Step 2: Reading beamcharge and daq LT generated by my scripts

```python
# Define the appropriate input variables
sbsconfch = [4, 7, 11, 14, 8, 9]
targetch = 'LD2'
replay_pass = 2
outfilech = '{}/all_beamCh_n_daqLvTm_prun_{}.csv'.format(targetch,targetch)
```

```python
# Create a list with all relevant files to be concatenated
prefixch  = '{}/beamCh_n_daqLvTm_prun_SBS'.format(targetch)
sheetsch = []    # list to store dfs per config 
for conf in sbsconfch:
    filename = prefixch + str(conf) + '_' + targetch + '.csv'
    tempdf = pd.DataFrame(pd.read_csv(filename))
    sheetsch.append(tempdf)
```

```python
# concatenate tables from all different sbs configs
concatdfch = pd.concat(sheetsch)
print(concatdfch.head())
print(concatdfch.shape)
```

```python
# check for nulls
anyNaN = concatdfch.isnull().values.any()
#final_table_nz_nd.loc[pd.isna(final_table_nz_nd).any(1), :].index
#if (anyNaN == True): 
#    print('NaN exists! Replacing with 0')
#    concatdfch = concatdfch.fillna(0)
print(anyNaN)
```

### Step 3: Concatenate the two

```python
# Finally merge the new tables with good run list
bob_n_pd = pd.merge(bobch, concatdfch, on='runnum', how='inner')
bob_n_pd
```

```python
# check for nulls
anyNaN_tot = bob_n_pd.isnull().values.any()
#final_table_nz_nd.loc[pd.isna(final_table_nz_nd).any(1), :].index
#if (anyNaN == True): 
#    print('NaN exists! Replacing with 0')
#    concatdfch = concatdfch.fillna(0)
print(anyNaN_tot)
```
